```python
# As we are only interested in the red channel - this snippet determines the camera filter pattern so we can extract 
# only the red channel digital number

bayer = raw.raw_image
bayer_pattern = ''.join(np.array(list((raw.color_desc).decode('ascii')))[raw.raw_pattern].flatten())
color = bayer_pattern.lower()

plt.figure(figsize=(2,2))
gs1 = gridspec.GridSpec(2,2)
gs1.update(wspace=0.05, hspace=0.05)
plt.suptitle('Bayer Pattern', fontsize=14)

for i in range(4):
    ax1 = plt.subplot(gs1[i])
    ax1.set_facecolor(color[i])
    ax1.xaxis.set_visible(False)
    ax1.yaxis.set_visible(False)
    ax1.set_xticklabels([])
    ax1.set_yticklabels([])
    ax1.set_aspect('equal')
    
plt.show()    
```

```python
# The image above tells us the red channel position.  From here, we use the following conditional to extract red and green 
# channels separately = []
red = []

for i in range(bayer.shape[0]):
    if i % 2 != 0:
        red.append(bayer[i][1::2])

green1, green2 = [],[]

for i in range(bayer.shape[0]):
    if i % 2 == 0:
        green1.append(bayer[i][1::2])
    else:
        green2.append(bayer[i][0::2])
```

```python
# Post pixel pushing to create new image capable files after extracting
red = np.concatenate(red, axis=0)
red.shape = (int(bayer.shape[0]/2), int(bayer.shape[1]/2))

green1 = np.concatenate(green1, axis=0)
green1.shape = (int(bayer.shape[0]/2), int(bayer.shape[1]/2))

green2 = np.concatenate(green2, axis=0)
green2.shape = (int(bayer.shape[0]/2), int(bayer.shape[1]/2))

green = (green1+green2)/2
green = green.astype(int)
```

```python
def green_halfres_stack_then_crop(raw_stack, x1, y1, wx1, wy1, bayer_pattern, out_dtype=np.float32):
    """
    raw_stack: (N,H,W) Bayer mosaic frames
    bayer_pattern: 4-char string in raster order of 2x2 tile, e.g. "RGGB" from rawpy

    Returns: (N, wy1, wx1) where the underlying green image is HALF-res (H//2, W//2),
             i.e. matches your original green1/green2 reshape coordinate system.
    """
    arr = np.asarray(raw_stack).astype(out_dtype, copy=False)  # accept list or array
    N, H, W = arr.shape

    tile = bayer_pattern.upper()
    c00, c01, c10, c11 = tile[0], tile[1], tile[2], tile[3]

    # find the two green positions in the 2x2 tile
    gpos = []
    if c00 == "G": gpos.append((0, 0))
    if c01 == "G": gpos.append((0, 1))
    if c10 == "G": gpos.append((1, 0))
    if c11 == "G": gpos.append((1, 1))

    (g1_r0, g1_c0), (g2_r0, g2_c0) = gpos[0], gpos[1]

    # Extract the two green lattices at half resolution
    g1 = arr[:, g1_r0::2, g1_c0::2]  # (N, H//2, W//2)
    g2 = arr[:, g2_r0::2, g2_c0::2]  # (N, H//2, W//2)

    green_half = 0.5 * (g1 + g2)     # (N, H//2, W//2)

    # Now crop in HALF-res coordinates (matches your original workflow)
    return green_half[:, y1:y1+wy1, x1:x1+wx1]


# ---- your geometry (kept exactly the same) ----
w1, w2 = 400, 400
x1, y1, wx1, wy1 = 790, 550, w1, w2

# usage:
# raw_arr = np.stack(raw_frames_list, axis=0)  # if you still have a list of 5 arrays
# green_roi_stack = green_halfres_stack_then_crop(raw_arr, x1, y1, wx1, wy1, bayer_pattern)


raw_arr = np.stack(raw_arr, axis=0)

green_roi_stack = green_halfres_stack_then_crop(
    raw_arr, x1, y1, wx1, wy1, bayer_pattern
)

np.save("green_roi_stack.npy", green_roi_stack)
```

```python
# Quantify Curvature, Banding, and High-Frequency Noise Quantified

H, W = green[y1:y1+wy1, x1:x1+wx1].shape
mean_dn = green[y1:y1+wy1, x1:x1+wx1].mean()

# --- CURVATURE (2-D low-frequency illumination gradient): quadratic surface fit ---
# z(x,y) = c0 + c1*x + c2*y + c3*x^2 + c4*x*y + c5*y^2

yy, xx = np.mgrid[0:H, 0:W]  # yy: rows, xx: cols

A = np.column_stack([
    np.ones(H * W),
    xx.ravel(),
    yy.ravel(),
    (xx**2).ravel(),
    (xx * yy).ravel(),
    (yy**2).ravel()
])

b = green[y1:y1+wy1, x1:x1+wx1].ravel()

coef, *_ = np.linalg.lstsq(A, b, rcond=None)

curv_field = (coef[0]
              + coef[1] * xx + coef[2] * yy
              + coef[3] * xx**2 + coef[4] * xx * yy + coef[5] * yy**2)

# Detrend while preserving overall brightness, then shift for visualization 
resid = green[y1:y1+wy1, x1:x1+wx1] - curv_field
im_detrended = (resid - resid.min())+300

# Keep using the existing plotting / flat-field visualization
flat_field.flat_field(im_detrended, 'A', 'B', 'uncurved.png')
```

```python
margin, win = 100, 50
centroids = {
    "center":      (W//2, H//2),
    "right_lower": (W - margin, margin),
    "right_upper": (W - margin, H - margin),
    "left_lower":  (margin, margin),
    "left_upper":  (margin, H - margin),
}

print(f"2D quadratic curvature metric on {win}x{win} windows (centroids inset by {margin}px, p95–p5 of fit):")
for name in ["center","right_lower","right_upper","left_lower","left_upper"]:
    cx, cy = centroids[name]
    rng, pct, (x0, y0) = Curvature.analyze(green[y1:y1+wy1, x1:x1+wx1], cx, cy, win, margin)
    print(f"  {name:12s} centroid (cx={cx:3d}, cy={cy:3d}) "
          f"-> patch @ (x0={x0:3d}, y0={y0:3d}):  {rng:6.2f} DN  ({pct:5.2f}% of mean)")
```

```python
def plot_windows_on_image_outline(W=400, H=400, win, margin):
    """
    Draw the 400x400 outline and the 5 analysis windows (rectangles) at:
      - center
      - right_upper
      - right_lower
      - left_upper
      - left_lower

    Windows are positioned by centroid, then converted to top-left (x0,y0),
    and clamped to stay in-bounds.
    """

    def top_left_from_centroid(cx, cy, win):
        x0 = int(round(cx - win/2))
        y0 = int(round(cy - win/2))
        x0 = int(np.clip(x0, 0, W - win))
        y0 = int(np.clip(y0, 0, H - win))
        return x0, y0

    centroids = {
        "center":      (W//2, H//2),
        "right_lower": (W - margin, margin),
        "right_upper": (W - margin, H - margin),
        "left_lower":  (margin, margin),
        "left_upper":  (margin, H - margin),
    }

    # Build window rectangles (x0,y0,w,h)
    windows = {}
    for name, (cx, cy) in centroids.items():
        x0, y0 = top_left_from_centroid(cx, cy, win)
        windows[name] = (x0, y0, win, win, cx, cy)

    fig, ax = plt.subplots(figsize=(6.5, 6.5))
    ax.set_aspect("equal", adjustable="box")

    # 400x400 outline
    ax.add_patch(Rectangle((0, 0), W, H, fill=False, linewidth=2))

    # Draw each window as its own box
    for name, (x0, y0, ww, hh, cx, cy) in windows.items():
        ax.add_patch(Rectangle((x0, y0), ww, hh, fill=False, linewidth=2))
        ax.plot(cx, cy, marker="o")  # centroid marker
        ax.text(x0 - 15 , y0 - 4, name, fontsize=9, va="top")

    ax.set_xlim(-10, W + 10)
    ax.set_ylim(-10, H + 10)
    ax.set_xlabel("x (pixels)")
    ax.set_ylabel("y (pixels)")
    ax.set_title(f"5 analysis windows (win={win}, margin={margin}) in a {W}x{H} image")

    plt.show()

# Example:
plot_windows_on_image_outline(W=400, H=400, win=win, margin=margin)
```

```python
path = Path("stack_column.csv")

# Read numeric values, ignoring comment lines like "# profile"
df = pd.read_csv(path, header=None, comment="#")

# If there are multiple columns, take the first non-empty numeric column
arr = df.apply(pd.to_numeric, errors="coerce").values
# Flatten and drop NaNs
profile = arr.ravel()
profile = profile[~np.isnan(profile)].astype(float)

N = profile.size
x = np.arange(N)

win = 30
margin = 100
deg = 2

coeffs = np.polyfit(x, profile, deg=deg)
lf = np.polyval(coeffs, x)


centers = {
    "upper left": margin,
    "center": N // 2,
    "upper right": N - margin
}

def window_bounds(c, win, N):
    x0 = int(round(c - win / 2))
    x0 = max(0, min(N - win, x0))
    return x0, x0 + win

windows = {name: window_bounds(c, win, N) for name, c in centers.items()}

metrics = {}
for name, (x0, x1) in windows.items():
    seg = lf[x0:x1]
    p5 = float(np.percentile(seg, 5))
    p95 = float(np.percentile(seg, 95))
    span = p95 - p5
    metrics[name] = {"x0": x0, "x1": x1, "p5": p5, "p95": p95, "span": span, "half": span/2}
```

```python
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from matplotlib.patches import Rectangle, ConnectionPatch
from matplotlib.ticker import FormatStrFormatter


fig, ax = plt.subplots(figsize=(11, 7), constrained_layout=True)

# --- Main plot (NO arrows) ---
ax.plot(x, profile, linewidth=1, label="Raw 1D profile (HF + LF)")
ax.plot(x, lf, linewidth=2, label=f"Low-frequency mean curve (poly deg={deg})")

# Shade all three ROIs on the main plot (optional but useful)
for name, m in metrics.items():
    x0, x1 = m["x0"], m["x1"]
    ax.axvspan(x0, x1-1, alpha=0.15)

# (b) Main plot y-range
ax.set_ylim(0.8, 1.0)

ax.set_xlabel("\nx (pixel column index)", fontsize=14)
ax.set_ylabel("Normalized Digital Number\n", fontsize=14)
ax.legend(loc="best", fontsize=14)
ax.tick_params(axis='both', labelsize=14)
ax.locator_params(axis='y', nbins=4)
ax.grid(
    True,
    which="both",
    linestyle="--",
    linewidth=0.5,
    alpha=0.4
)

# --- (c) Inset: left ROI with arrows ---
# pick which ROI to inset
roi_name = "upper left"
m = metrics[roi_name]
x0, x1 = m["x0"], m["x1"]

# Create inset axes (tweak width/height/location as you like)
axins = inset_axes(
    ax,
    width="100%", height="100%",
    bbox_to_anchor=(0.4, 0.4, 0.4, 0.4),  # (x0, y0, w, h)
    bbox_transform=ax.transAxes,
    borderpad=0
)

# Plot only the left ROI region in the inset
axins.plot(x[x0:x1], profile[x0:x1], linewidth=1)
axins.plot(x[x0:x1], lf[x0:x1], linewidth=2)

# --- Highlight upper-left ROI on main plot ---
roi_name = "upper left"
m = metrics[roi_name]
x0, x1 = m["x0"], m["x1"]

# ROI y-range from data inside the ROI
roi_ymin = min(profile[x0:x1].min(), lf[x0:x1].min())
roi_ymax = max(profile[x0:x1].max(), lf[x0:x1].max())

# Add padding (e.g., 15% of the ROI span)
pad = 0.5 * (roi_ymax - roi_ymin) if roi_ymax > roi_ymin else 0.01
y0_box = roi_ymin - pad
y1_box = roi_ymax + pad


roi_box = Rectangle(
    (x0, y0_box),
    width=(x1 - x0),
    height=(y1_box - y0_box),
    fill=False,
    edgecolor="black",
    linewidth=2
)


ax.add_patch(roi_box)


# Auto y-limits based on data in the inset window (pad a bit)
ymin = min(profile[x0:x1].min(), lf[x0:x1].min())
ymax = max(profile[x0:x1].max(), lf[x0:x1].max())
pad = 0.10 * (ymax - ymin) if ymax > ymin else 0.01
axins.set_ylim(ymin - pad, ymax + pad)

# Add p5–p95 arrow on the LF curve *within this ROI*
lx = x0 + 1  # x-position for arrow (still in full x coords)
axins.annotate(
    "",
    xy=(lx, m["p95"]),
    xytext=(lx, m["p5"]),
    arrowprops=dict(arrowstyle="<->", linewidth=3, color='r')
)

# Optional: label the span in the inset
axins.text(
    0.08, 0.65,
    f"p95 = {m['p95']:.3f}",
    transform=axins.transAxes,
    fontsize=10,
    va="top",
    bbox=dict(
        facecolor="white",
        edgecolor="none",
        alpha=0.85
    )
)

axins.text(
    0.08, 0.3,
    f"p5  = {m['p5']:.3f}",
    transform=axins.transAxes,
    fontsize=10,
    va="bottom",
    bbox=dict(
        facecolor="white",
        edgecolor="none",
        alpha=0.85
    )
)



# Make inset ticks readable but slightly smaller

axins.tick_params(axis='both', labelsize=12)
axins.yaxis.set_major_formatter(FormatStrFormatter('%.2f'))
axins.locator_params(axis='y', nbins=4)
axins.grid(
    True,
    which="both",
    linestyle="--",
    linewidth=0.5,
    alpha=0.4
)


# (optional) tighten inset x-limits exactly to the ROI
axins.set_xlim(x0, x1-1)

# Pick a point on the ROI box (top-right corner)
roi_point = (x1, y0_box)

# Pick a point in the inset (top-left corner in inset coords)
inset_point = (0, 1)

con = ConnectionPatch(
    xyA=inset_point, coordsA=axins.transAxes,
    xyB=roi_point,   coordsB=ax.transData,
    arrowstyle="-",
    linewidth=2,
    color="black"
)

fig.add_artist(con)


out_png = "curvature_1d_demo.png"
fig.savefig(out_png, dpi=300)
```
